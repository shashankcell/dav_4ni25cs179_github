```python
import numpy as np
import pandas as pd

data = {
    "name": ["amit", "bebo", "chetan", "amit", "diya", "ravi"],
    "age": [25, np.nan, 30, 25, 28, 22],
    "marks": [85, 90, np.nan, 85, 78, 95],
    "city": ["Mysuru", "Mysuru", "Bengaluru", "Mysuru", np.nan, "Delhi"]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("Shape of dataset(rows, col)")
```

```python
print("\nShape of dataset (rows, columns):", df.shape)

print("\nColumn names:", df.columns.tolist())

print("\nData types of each column:")
print(df.dtypes)

print("\nDataset info:")
df.info()
```

```python
print("\nDescriptive statistics (numerical columns):")
print(df.describe())

print("\nMean:", df.mean(numeric_only=True))

print("\nMedian:", df.median(numeric_only=True))

print("\nStandard Deviation:", df.std(numeric_only=True))
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python
print("\nDuplicate rows in dataset:")
print(df[df.duplicated()])
print("\nNumber of duplicate rows:", df.duplicated().sum())
print("\nEDA process completed succesfully!")
```
