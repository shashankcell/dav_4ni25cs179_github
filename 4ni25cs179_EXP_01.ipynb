```python


import numpy as np
import pandas as pd

marks = np.array([75, 85, 95, 88])

print("Marks:", marks)
print("Type:", type(marks))
print("Number of dimension:", marks.ndim)
print("Shape:", marks.shape)
print("Number of elements:", marks.size)
```

```python


print("Original:", marks)
print("Add 5:", marks + 5)

print("Sum:", marks.sum())
print("Mean:", marks.mean())
print("Maximum:", marks.max())
```

```python


print("First marks:", marks[0])
print("Second marks:", marks[2])
print("Last marks:", marks[-1])
```

```python


print("Index 1 to 3:", marks[1:4])
print("First three:", marks[:3])
print("From index 2:", marks[2:])
```

```python


data = {
    "Name": ["Anu", "Rahul", "Kiran", "Priya"],
    "Branch": ["CSE", "AIML", "CSE", "AIML"],
    "Marks": [85, 91, 76, 89]
}

df = pd.DataFrame(data)
df
```

```python


print("Shape:", df.shape)
print("Column:", df.columns)
```

```python


df.info()
```

```python


df.tail()
```

```python


print("First row:")
print(df.iloc[0])

print("Second row:\n", df.iloc[1])

print("First three row:\n", df.iloc[:3])

print("Last two row:\n", df.iloc[-2:])

print("First two row and first two columns:\n", df.iloc[:2, :2])
```

```python


print("Students from CSE:", df[df["Branch"] == "CSE"])

print("Students scoring 85 or above:", df[df["Marks"] >= 85])
```
